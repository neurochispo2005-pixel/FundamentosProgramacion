Ejercicio 10: Suma acumulada con 'while' 

Pide números mientras sean distintos de `0` (el `0` termina el ciclo y no se suma).

```python
suma = 0                            # Contador

while True:                         # Suma acumulada
    numero = int(input("ingrese un numero entero, si desea terminar, ingrese 0: "))
    print (numero)

    if numero == 0:              # Break para terminar ciclo
        break

    suma += numero
print("la suma de todos los numeros es: ", suma)
```

Ejercicio 12: Suma de pares del 1 al 20 con `for` y `continue`



```python
suma = 0                               # Contador

for numero in range(1, 21):            # For para suma de pares del 1 al 20
    if numero % 2 != 0:
        continue 
    suma += numero
print("la suma de los numeros pares del 1 al 20 es: ", suma)
```

Ejercicio 13: Suma de una lista con función que regresa valor (sin `sum()`)

```python
def sumar_lista(lista):              # Creación de función sumar_lista
    total = 0
    for elemento in lista:           # Recorrer cada elemento
        total += elemento
    return total

numeros = [10, 20, 30, 40]           # Lista del ejercicio
resultado = sumar_lista(numeros)
print("la suma de todos los elementos es: ",resultado)
```

Ejercicio 14: Máximo de una lista con función que regresa valor (sin `max()`)

```python
def obtener_maximo(lista):           # Creación de función obtener_maximo
    maximo = lista[0]
    for numero in lista[1:]:        
        if numero > maximo:
            maximo = numero
    return maximo

numeros = [7, 3, 9, 2]               # Lista del ejercicio
resultado = obtener_maximo(numeros)
print("el elemento maximo es: ",resultado)
```

Ejercicio 15: Matriz y suma de la diagonal principal

```python
matriz = [                     # Creación de matriz
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
    ]

suma_diagonal = 0             # Contador
for s in range (len(matriz)):
    suma_diagonal += matriz [s][s]         # Suma de elementos de la matriz 
print ("la suma de la diagonal principal es: ",suma_diagonal)
```

Ejercicio 16: Promedio con función que no regresa valor

```python
def calcular_promedio(lista):          # Creación función calcular_promedio
    
    suma = 0                           # Contador

    for elemento in lista:             # Recorrer elementos de la lista
        suma += elemento
    promedio = suma / len(lista)       # Promedio

    print("el promedio es: ",promedio)

calificaciones = [8, 9, 10, 7]          # Calificaciones del ejercicio
calcular_promedio(calificaciones)
```

Ejercicio 18: Factorial con función recursiva

```python
def factorial(n):                      # Creación función factorial
    if n == 0 or n == 1:               # Numero de 0 y 1 con condición
        return 1   
    else:
        return n * factorial (n - 1)    # de lo contrario, hacer el factorial

numero = int(input("ingrese el numero que desee: "))
resultado = factorial(numero)
print(f"el factorial de {numero} es: {resultado}")
```

Ejercicio 19: Tupla: acceso y suma con función

```python
def procesar_tupla(tupla):                 # Creación función procesar_tupla

    tercer_elemento = tupla[2]             # Identificar el tercer elemento
    suma_total = 0
    for numero in tupla:
        suma_total += numero
        
    return tercer_elemento, suma_total

numeros = (4, 8, 15, 16, 23, 42)           # Números del ejercicio
tercer_elemento, suma = procesar_tupla(numeros)   # Operación
print(f"el tercer elemento es: {tercer_elemento}")
print(f"la suma de la tupla es: {suma}")
```

Ejercicio 20: Diccionario: buscar teléfono de un contacto

```python
contactos = {                      # Creación del diccionario de los contactos
    "beto": "5555678",
    "fatima": "4429876543",
    "aldo": "4425550199"
}

def buscar_contacto(contactos, nombre):   # Creación función buscar_contactos
    return contactos.get(nombre, "contacto no encontrado")   # Identificar contacto

nombre = input("escribe el nombre del contacto que desee encontrar: ")
telefono = buscar_contacto(contactos, nombre)
print(f"el teléfono de {nombre} es: {telefono}")
```
