# Bloque 1: Decisiones y operadores lógicos

Ejercicio 1: Tarifa de estacionamiento

Un estacionamiento cobra las primeras 2 horas a $15.00 cada una y
las horas adicionales a $10.00 cada una. El cobro nunca puede exceder los $200.00
del pase diario (tope). Crea un programa que pida el número de horas (entero) y
muestre el total a pagar aplicando las reglas.

```python
# Pedimos al usuario el número de horas
horas = int(input("Ingresa el número de horas: "))

# Si son 2 horas o menos, cada hora cuesta $15
if horas <= 2:
    total = horas * 15

# Si son más de 2 horas, las primeras 2 cuestan $15
# y las horas restantes cuestan $10
else:
    total = (2 * 15) + ((horas - 2) * 10)

# Revisamos si el total supera el límite de $200
if total > 200:
    total = 200

# Mostramos el total a pagar con dos decimales
print("Total a pagar: ${:.2f}".format(total))
```

Ejercicio 2: años bisiestos en un rango

Crea un programa que pida dos años (inicio y fin) y muestre en una
línea todos los años bisiestos de ese rango. Regla: un año es bisiesto si es
divisible entre 4 pero no entre 100, o si es divisible entre 400. Usa el `for` con
`range` y operadores lógicos (`and`/`or`).

```python
# Pedimos el año donde comienza el rango
inicio = int(input("Ingresa el año inicial: "))

# Pedimos el año donde termina el rango
fin = int(input("Ingresa el año final: "))

# Mostramos el texto antes de empezar a imprimir los años
print("Años bisiestos:", end=" ")

# Recorremos todos los años desde el inicio hasta el final
for año in range(inicio, fin + 1):

    # Revisamos si el año cumple con las reglas para ser bisiesto
    if (año % 4 == 0 and año % 100 != 0) or año % 400 == 0:

        # Mostramos el año que sí es bisiesto
        print(año, end=" ")
```

Ejercicio 3: Contador de dígitos

Crea un programa que pida un número entero positivo y cuántos
dígitos tiene, usando `while` con el operador división entera `//` (cada iteración
elimina el último dígito).

```python
# Pedimos un número entero positivo
numero = int(input("Ingresa un número entero positivo: "))

# Guardamos el número original para mostrarlo al final
original = numero

# Creamos un contador que empieza en cero
contador = 0

# Repetimos mientras todavía queden dígitos en el número
while numero > 0:

    # Dividimos entre 10 usando división entera
    # Esto elimina el último dígito
    numero = numero // 10

    # Aumentamos el contador porque eliminamos un dígito
    contador = contador + 1

# Mostramos la cantidad de dígitos que tenía el número
print("El número", original, "tiene", contador, "dígitos")
```

Ejercicio 4: Suma de dígitos

Crea un programa que pida un número entero positivo y calcule la
suma de sus dígitos usando `while`, el módulo `%` para extraer el último dígito y
`//` para eliminarlo.

```python
# Pedimos un número entero positivo
numero = int(input("Ingresa un número entero positivo: "))

# Guardamos el número original para mostrarlo al final
original = numero

# Creamos una variable para acumular la suma
suma = 0

# Repetimos mientras el número todavía tenga dígitos
while numero > 0:

    # Sacamos el último dígito usando el módulo
    digito = numero % 10

    # Sumamos el dígito al acumulador
    suma = suma + digito

    # Eliminamos el último dígito usando división entera
    numero = numero // 10

# Mostramos el resultado de la suma
print("La suma de los dígitos de", original, "es:", suma)
```

Ejercicio 5: Número invertido

Crea un programa que pida un número entero positivo y muestre sus
dígitos en orden inverso. Construye el nuevo número en un acumulador con la fórmula
`invertido = invertido * 10 + digito`.

```python
# Pedimos un número entero positivo
numero = int(input("Ingresa un número entero positivo: "))

# Guardamos el número original para mostrarlo al final
original = numero

# Creamos el acumulador del número invertido
invertido = 0

# Repetimos mientras todavía existan dígitos
while numero > 0:

    # Sacamos el último dígito del número
    digito = numero % 10

    # Agregamos el dígito al número invertido
    invertido = invertido * 10 + digito

    # Eliminamos el último dígito del número original
    numero = numero // 10

# Mostramos el número invertido
print("El número invertido de", original, "es", invertido)
```

Ejercicio 6: División con restas sucesivas

Sin usar los operadores `/` ni `//`, calcula el cociente y el
residuo de una división entre enteros positivos restando el divisor al dividendo
con un `while`; el cociente es la cantidad de restas realizadas.

```python
# Pedimos el dividendo
dividendo = int(input("Ingresa el dividendo: "))

# Pedimos el divisor
divisor = int(input("Ingresa el divisor: "))

# Guardamos el dividendo original para mostrarlo al final
original = dividendo

# Creamos el contador del cociente
cociente = 0

# Repetimos mientras podamos restar el divisor
while dividendo >= divisor:

    # Restamos el divisor al dividendo
    dividendo = dividendo - divisor

    # Cada resta representa una unidad del cociente
    cociente = cociente + 1

# Lo que queda después de las restas es el residuo
residuo = dividendo

# Mostramos el resultado de la división
print(original, "÷", divisor, "=", cociente, "con residuo", residuo)
```

Ejercicio 8: Potencia sin usar **

Crea un programa que pida una base y un exponente entero positivo, y
calcule la potencia mediante multiplicaciones sucesivas con un `for` y un
acumulador (sin usar el operador `**`).

```python
# Pedimos la base
base = int(input("Ingresa la base: "))

# Pedimos el exponente
exponente = int(input("Ingresa el exponente: "))

# Empezamos el resultado en 1
resultado = 1

# Repetimos la multiplicación la cantidad de veces indicada
for i in range(exponente):

    # Multiplicamos el resultado por la base
    resultado = resultado * base

# Mostramos el resultado de la potencia
print(base, "elevado a", exponente, "=", resultado)
```

Ejercicio 9: El mayor N números

Crea un programa que primero pregunte cuántos números se capturarán
y, con un `for`, lea cada uno guardando siempre el mayor visto hasta ese momento en
una variable.

```python
# Preguntamos cuántos números se van a ingresar
cantidad = int(input("¿Cuántos números vas a ingresar? "))

# Creamos la variable para guardar el número mayor
mayor = 0

# Repetimos el proceso la cantidad de veces indicada
for i in range(cantidad):

    # Pedimos un número
    numero = int(input("Ingresa un número: "))

    # Si es el primer número, lo tomamos como el mayor
    if i == 0:
        mayor = numero

    # Si el número nuevo es mayor, lo guardamos
    elif numero > mayor:
        mayor = numero

# Mostramos el número mayor
print("El número mayor es:", mayor)
```

Ejercicio 11: Serie alternada

Crea un programa que pida un entero positivo `n` y calcule la suma
alternada `1 - 2 + 3 - 4 + ...` hasta `n`. Usa un `for` y decide el signo de cada
término con el módulo `%`.

```python
# Pedimos hasta qué número llegará la serie
n = int(input("Ingresa un número positivo: "))

# Creamos el acumulador de la suma
suma = 0

# Recorremos los números desde 1 hasta n
for numero in range(1, n + 1):

    # Revisamos si el número es par
    if numero % 2 == 0:

        # Si es par, lo restamos
        suma = suma - numero

    # Si no es par, significa que es impar
    else:

        # Si es impar, lo sumamos
        suma = suma + numero

# Mostramos el resultado de la serie
print("El resultado de la serie es:", suma)
```

Ejercicio 12: Serie armónica

Crea un programa que pida un entero positivo `n` y calcule la suma
`1 + 1/2 + 1/3 + ... + 1/n` con un acumulador flotante. Muestra el resultado con 2
decimales.

```python
# Pedimos hasta qué número llegará la serie
n = int(input("Ingresa un número positivo: "))

# Creamos un acumulador decimal
suma = 0.0

# Recorremos los números desde 1 hasta n
for numero in range(1, n + 1):

    # Sumamos 1 dividido entre el número actual
    suma = suma + (1 / numero)

# Mostramos el resultado con solamente 2 decimales
print("La suma de la serie es: {:.2f}".format(suma))
```

Ejercicio 13: Sucesión de Fibonacci

Crea un programa que pida cuántos términos de la sucesión de
Fibonacci mostrar. La sucesión comienza con `0` y `1`, y cada término siguiente es
la suma de los dos anteriores. Usa dos variables que avancen junto con el `for`.

```python
# Preguntamos cuántos términos quiere mostrar el usuario
n = int(input("¿Cuántos términos quieres mostrar? "))

# Primer número de Fibonacci
a = 0

# Segundo número de Fibonacci
b = 1

# Mostramos el texto inicial
print("Fibonacci:", " ")

# Repetimos según la cantidad de términos solicitada
for i in range(n):

    # Mostramos el número actual
    print(a, " ")

    # Calculamos el siguiente número sumando los dos anteriores
    siguiente = a + b

    # Movemos las variables para continuar con la sucesión
    a = b
    b = siguiente
```

Ejercicio 14: Triángulo de asteriscos

Crea un programa que pida un número de filas y dibuje un triángulo
de asteriscos cuya primera fila tenga 1 asterisco, la segunda 2, y así
sucesivamente. Usa ciclos anidados.

```python
# Pedimos la cantidad de filas
filas = int(input("Ingresa el número de filas: "))

# Recorremos las filas desde 1 hasta el número indicado
for i in range(1, filas + 1):

    # Creamos una cadena vacía para guardar los asteriscos
    linea = ""

    # Repetimos según el número de la fila
    for i in range(i):

        # Agregamos un asterisco a la cadena
        linea = linea + "*"

    # Mostramos la fila completa
    print(linea)
```
