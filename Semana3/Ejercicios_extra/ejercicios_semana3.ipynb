Ejercicio 6: Calificación con escala de letras

Crea un programa que pida una calificación numérica (0 a 10) y la
convierta a letra: `A` si es `>= 9`, `B` si es `>= 8`, `C` si es `>= 7`, `D` si es
`>= 6`, `F` en otro caso.

```python
# Pedimos la calificación
calificacion = float(input("Ingresa tu calificación: "))

# Revisamos si la calificación es mayor o igual a 9
if calificacion >= 9:

    # La calificación corresponde a A
    letra = "A"

# Revisamos si es mayor o igual a 8
elif calificacion >= 8:

    # La calificación corresponde a B
    letra = "B"

# Revisamos si es mayor o igual a 7
elif calificacion >= 7:

    # La calificación corresponde a C
    letra = "C"

# Revisamos si es mayor o igual a 6
elif calificacion >= 6:

    # La calificación corresponde a D
    letra = "D"

# Si no cumple ninguna condición
else:

    # La calificación corresponde a F
    letra = "F"

# Mostramos la letra obtenida
print("Tu calificación es:", letra)
```

Ejercicio 9: Descuento escalonado

Crea un programa que aplique un descuento según el monto de compra:
sin descuento si es menor a $100, 10% si es de $100 a $499, y 20% si es de $500 o
más. Muestra el precio final con dos decimales.

```python
# Pedimos el monto de la compra
monto = float(input("Ingresa el monto de la compra: "))

# Si el monto es menor a 100, no hay descuento
if monto < 100:

    # El descuento es 0
    descuento = 0

# Si el monto es menor a 500, se aplica 10%
elif monto < 500:

    # Calculamos el 10% de descuento
    descuento = monto * 0.10

# Si el monto es de 500 o más
else:

    # Calculamos el 20% de descuento
    descuento = monto * 0.20

# Restamos el descuento al monto
final = monto - descuento

# Mostramos el precio final con dos decimales
print("Precio final con descuento: ${:.2f}".format(final))
```

Ejercicio 10: Tipo de triángulo

Crea un programa que pida tres lados y determine si el triángulo es
equilátero (tres lados iguales), isósceles (dos iguales) o escaleno (todos
distintos).

```python
# Pedimos los tres lados
lado1 = float(input("Ingresa el primer lado: "))
lado2 = float(input("Ingresa el segundo lado: "))
lado3 = float(input("Ingresa el tercer lado: "))

# Revisamos si los tres lados son iguales
if lado1 == lado2 and lado2 == lado3:

    # Si son iguales, es equilátero
    print("El triángulo es equilátero")

# Revisamos si por lo menos dos lados son iguales
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:

    # Si dos lados son iguales, es isósceles
    print("El triángulo es isósceles")

else:

    # Si todos son diferentes, es escaleno
    print("El triángulo es escaleno")
```

Ejercicio 14: Menú simple con while

Crea un programa con un menú de opciones: `1` sumar, `2` restar, `3`
salir. Para las opciones 1 y 2 pide dos números y muestra el resultado; repite
hasta elegir salir.

```python
# Creamos una opción inicial
opcion = 0

# Repetimos mientras no se elija salir
while opcion != 3:

    # Mostramos el menú
    print("\n--- MENÚ ---")
    print("1. Sumar")
    print("2. Restar")
    print("3. Salir")

    # Pedimos una opción
    opcion = int(input("Selecciona una opción: "))

    # Si elige sumar
    if opcion == 1:

        # Pedimos dos números
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        # Realizamos la suma
        resultado = numero1 + numero2

        # Mostramos el resultado
        print("Resultado:", resultado)

    # Si elige restar
    elif opcion == 2:

        # Pedimos dos números
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        # Realizamos la resta
        resultado = numero1 - numero2

        # Mostramos el resultado
        print("Resultado:", resultado)

    # Si elige salir
    elif opcion == 3:

        # Mostramos el mensaje final
        print("Fin del programa")

    else:

        # Mensaje cuando la opción no existe
        print("Opción no válida")
```

Ejercicio 15: Tabla de multiplicar con while

Crea un programa que pida un número y muestre su tabla de
multiplicar del 1 al 10 usando un ciclo `while`.

```python
# Pedimos el número de la tabla
numero = int(input("Ingresa un número: "))

# Comenzamos el contador en 1
contador = 1

# Repetimos hasta llegar al 10
while contador <= 10:

    # Calculamos la multiplicación
    resultado = numero * contador

    # Mostramos la operación
    print(numero, "x", contador, "=", resultado)

    # Aumentamos el contador
    contador = contador + 1
```

Ejercicio 16: Suma de los primeros N números

Crea un programa que pida un entero positivo `n` y calcule la suma
de los números del 1 al `n` con un ciclo `for`.

```python
# Pedimos el número final
n = int(input("Ingresa un número positivo: "))

# Creamos el acumulador
suma = 0

# Recorremos los números desde 1 hasta n
for numero in range(1, n + 1):

    # Sumamos el número actual
    suma = suma + numero

# Mostramos el resultado
print("La suma de los primeros", n, "números es:", suma)
```

Ejercicio 17: Factorial con 'for'

Crea un programa que pida un entero positivo y calcule su factorial
(producto de 1 hasta N) usando un ciclo `for`.

```python
# Pedimos un número
n = int(input("Ingresa un número positivo: "))

# El factorial comienza en 1
factorial = 1

# Recorremos desde 1 hasta n
for numero in range(1, n + 1):

    # Multiplicamos el factorial por el número actual
    factorial = factorial * numero

# Mostramos el resultado
print("El factorial de", n, "es:", factorial)
```

Crea un programa que pida un entero positivo y realice una cuenta
regresiva desde ese número hasta `1`, imprimiendo los valores en una línea y el
mensaje de despegue al final.

```python
# Pedimos el número inicial
n = int(input("Ingresa un número positivo: "))

# Recorremos desde n hasta 1
for numero in range(n, 0, -1):

    # Mostramos el número en la misma línea
    print(numero, end="")

    # Ponemos una coma si todavía no llegamos al 1
    if numero > 1:
        print(", ", end="")

# Mostramos el mensaje final
print(", ¡Despegue!")
```

Ejercicio 21: Validación de rango  y tabla con 'for'

Crea un programa que pida un número entre 1 y 10; si está fuera de
rango, muestra un error y vuelve a pedirlo con `while`. Cuando sea válido, imprime
su tabla de multiplicar del 1 al 10 con `for`.

```python
# Pedimos un número
numero = int(input("Ingresa un número entre 1 y 10: "))

# Repetimos mientras el número esté fuera del rango
while numero < 1 or numero > 10:

    # Mostramos el error
    print("Error: número fuera de rango (1-10)")

    # Volvemos a pedir el número
    numero = int(input("Ingresa un número entre 1 y 10: "))

# Mostramos la tabla del número válido
for i in range(1, 11):

    # Calculamos la multiplicación
    resultado = numero * i

    # Mostramos la operación
    print(numero, "x", i, "=", resultado)
```

Ejercicio 22: Conteo de positivos, negativos y ceros

Crea un programa que pida cuántos valores se van a capturar y luego
los capture con `for`, contando cuántos son positivos, cuántos negativos y cuántos
ceros.

```python
# Preguntamos cuántos números se van a ingresar
cantidad = int(input("¿Cuántos números vas a ingresar? "))

# Creamos los contadores
positivos = 0
negativos = 0
ceros = 0

# Repetimos según la cantidad de números
for i in range(cantidad):

    # Pedimos un número
    numero = int(input("Ingresa un número: "))

    # Revisamos si es positivo
    if numero > 0:

        # Aumentamos el contador de positivos
        positivos = positivos + 1

    # Revisamos si es negativo
    elif numero < 0:

        # Aumentamos el contador de negativos
        negativos = negativos + 1

    else:

        # Si no es positivo ni negativo, es cero
        ceros = ceros + 1

# Mostramos los resultados
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Ceros:", ceros)
```

Ejercicio 23: Promedio con validación de rango

Crea un programa que capture calificaciones hasta ingresar `-1`. Si
una calificación está fuera del rango 0 a 10, muestra un error y no la considera.
Al final muestra el promedio de las válidas.

```python
# Creamos un acumulador para las calificaciones
suma = 0

# Creamos un contador para las calificaciones válidas
contador = 0

# Pedimos la primera calificación
calificacion = float(input("Ingresa una calificación (-1 para terminar): "))

# Repetimos mientras no se ingrese -1
while calificacion != -1:

    # Revisamos si la calificación está fuera del rango
    if calificacion < 0 or calificacion > 10:

        # Mostramos un mensaje de error
        print("Error: calificación fuera de rango (0-10)")

    else:

        # Sumamos la calificación válida
        suma = suma + calificacion

        # Aumentamos el contador
        contador = contador + 1

    # Pedimos otra calificación
    calificacion = float(input("Ingresa otra calificación (-1 para terminar): "))

# Calculamos el promedio de las calificaciones válidas
promedio = suma / contador

# Mostramos el promedio
print("El promedio es: {:.1f}".format(promedio))
```

Ejercicio 24: Minicalculadora con menu y 'while'

Crea un programa con un menú: `1` suma, `2` resta, `3`
multiplicación, `4` salir. Para cada operación pide dos números y muestra el
resultado; repite hasta elegir salir.

```python
# Creamos una opción inicial
opcion = 0

# Repetimos hasta que el usuario seleccione 4
while opcion != 4:

    # Mostramos el menú
    print("\n--- MINICALCULADORA ---")
    print("1. Suma")
    print("2. Resta")
    print("3. Multiplicación")
    print("4. Salir")

    # Pedimos una opción
    opcion = int(input("Selecciona una opción: "))

    # Opción para sumar
    if opcion == 1:

        # Pedimos los dos números
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        # Realizamos la suma
        resultado = numero1 + numero2

        # Mostramos el resultado
        print("Suma:", resultado)

    # Opción para restar
    elif opcion == 2:

        # Pedimos los dos números
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        # Realizamos la resta
        resultado = numero1 - numero2

        # Mostramos el resultado
        print("Resta:", resultado)

    # Opción para multiplicar
    elif opcion == 3:

        # Pedimos los dos números
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        # Realizamos la multiplicación
        resultado = numero1 * numero2

        # Mostramos el resultado
        print("Multiplicación:", resultado)

    # Opción para salir
    elif opcion == 4:

        # Mostramos el mensaje final
        print("Fin del programa")

    else:

        # Mostramos un mensaje si la opción no existe
        print("Opción no válida")
```
