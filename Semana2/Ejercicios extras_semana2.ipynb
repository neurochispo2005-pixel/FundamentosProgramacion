# Ejercicio extra 1 (Control de aforo con break y continue (while))
Crea un programa que procese visitantes del museo con un ciclo 'while'. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "menor de tres años, sin cargo" y continua con el siguiente visitante (usa 'continue'). Si el total acumulado alcanza o supera $100 detén el ciclo con 'break' y muestra el total.

```python
total = 0

while True:

    edad = int(input("Ingresa la edad del visitante: "))

    if edad < 3:
        print("Menor de 3 años, sin cargo")
        continue

    if edad >= 3 and edad <= 17:
        boleto = 30
    else:
        boleto = 45

    print("Costo del boleto: $", boleto)

    total = total + boleto

    print("Total acumulado: $", total)

    if total >= 100:
        print("aforo alcanzado, se detiene el registro")
        break

print("TOTAL A PAGAR: $", total)
print("Programa terminado.")
```

Ejercicio extra 2 (Estadística de visitantes con for)

Crea un programa que pida el número total de visitantes y, con un
ciclo `for`, capture la edad de cada uno. Debe contar cuántos son adultos (18 años
o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos
resultados con dos decimales.

```python
visitantes = int(input("Ingresa el numero total de visitantes: "))

adultos = 0
suma_edades = 0

for i in range(visitantes):

    edad = int(input("Ingresa la edad del visitante " + str(i + 1) + ": "))

    suma_edades = suma_edades + edad

    if edad >= 18:
        adultos = adultos + 1

promedio = suma_edades / visitantes
print("RESULTADOS")
print("Cantidad de adultos:", adultos)
print("Promedio de edad: {:2f}".format(promedio))
```

El siguiente código pretende calcular el precio final de un boleto de $45 con el descuento del adulto mayor (12%) pero contiene un error de lógica. Inserta 'pdb.set_trace()'para rastrear las variables, identifica el error y escribe la versión corregida

Ejercicio extra 4 (Pirámide de asteriscos con for)

Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos 'for' anidados y el parámetro 'end' de 'print'.

```python
altura  = int(input("Ingresa la altura de la piramide:"))

for i in range(1,altura+1):
    print("*"*i)
```
