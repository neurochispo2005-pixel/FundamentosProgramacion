Ejercicio extra 1: Suma todos los elementos de una matriz

Crea una función `suma_matriz(matriz)` que **regrese valor** (con
`return`) la suma de todos los elementos de una lista de listas, usando ciclos
`for` anidados. Define además una función `imprimir_matriz(matriz)` que **no
regrese valor** y muestre la matriz sin corchetes ni comas. Aplica ambas funciones
a la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]`.

```python
def suma_matriz(matriz):   # Creacion funcion suma_matriz
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma += elemento
    return suma

def imprimir_matriz(matriz):  # Creacion funcion imprimir_matriz
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

# Matriz
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

imprimir_matriz(matriz)    # Imprime matriz
resultado = suma_matriz(matriz)  # Obtener y mostrar la suma
print("La suma de todos los elementos de la matriz es:", resultado)
```

Ejercicio extra 2: Matriz con suma por fila

Crea una función `suma_fila(fila)` que **regrese valor** (con
`return`) la suma de los elementos de una lista, y una función
`mostrar_tabla(matriz)` que **no regrese valor** e imprima la matriz sin corchetes
ni comas. Dada la matriz `[[3, 1, 4], [1, 5, 9], [2, 6, 5]]`, imprime la matriz y
la suma de cada fila.


```python
def suma_fila(fila):   # Creacion funcion suma_fila
    suma = 0
    for elemento in fila:
        suma += elemento
    return(suma)


def mostrar_tabla(matriz):  # Creacion funcion mostrar_tabla
    for fila in matriz:
        for elemento in fila:
            print((elemento), end = "  ")
        print("")

# Matriz
matriz = [
    [3, 1, 4],
    [1, 5, 9],
    [2, 6, 5]
    ]

mostrar_tabla(matriz)   # Mostrar la matriz
suma = 0
for fila in matriz:     # Recorrer cada fila
    resultado = suma_fila(fila)
    suma = resultado
    print("")
    print ("La suma de la fila es: ", suma)
```

Ejercicio extra 3: Producto de una columna sin usar el operador *

Crea la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]` (4
filas y 3 columnas). Define una función `multiplicar_columna(matriz, columna)` que
**regrese valor** (con `return`) el producto de los elementos de una columna dada.
**Restricción:** dentro de la función está prohibido usar el operador `*`; el
producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el
número de columna (`0`, `1` o `2`) y muestra el resultado.

```python
def multiplicar_columna(matriz, columna):
    resultado = matriz[0][columna]      #  acumulador
    for fila in range(1, len(matriz)):
        numero = matriz[fila][columna]
        suma = 0                        # Multiplicación mediante sumas repetidas
        for contador in range(numero):
            suma += resultado
        resultado = suma
    return resultado

# Matriz
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    [2, 4, 6]
]

columna = int(input("Numero de columna: "))   # Capturar numero de columna
if columna >= 0 and columna <= 2:             # Validar columna
    resultado = multiplicar_columna(matriz, columna)
    print("El producto de la columna", columna, "es:", resultado)
else:
    print("La columna debe ser 0, 1 o 2.")
```

Ejercicio extra 4: Tabla de pitágoras de tamaño variable

Crea una función `generar_tabla(tamano)` que **regrese valor** (con
`return`) una lista de listas de `tamano x tamano` con los productos de la tabla de
Pitágoras (renglones y columnas del 1 al `tamano`). Dentro de la función de
generación está prohibido usar el operador `*`; usa sumas repetidas con ciclos.
Define una función `imprimir_tabla(tabla)` que **no regrese valor** e imprima la
tabla sin corchetes ni comas, con los números tabulados (`\t`). Pide al usuario un
tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de
error. Al final pide un renglón y una columna y muestra el producto consultando la
matriz, sin usar `*`.

```python
def generar_tabla(tamano):    # Creación funcion generar_tabla
    tabla = []
    for renglon in range(1, tamano + 1):   # Recorrer renglones
        fila = []
        for columna in range(1, tamano + 1):  # Recorrer columnas
            resultado = 0
            for contador in range(columna):    # Producto mediante sumas repetidas
                resultado += renglon
            fila.append(resultado)   # Agregar resultados
        tabla.append(fila)
    return tabla

def imprimir_tabla(tabla):   # Creacion funcion imprimir_tabla
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t")
        print()

tamano = int(input("Tamano de la tabla: "))   # Pedir tamaño

if tamano < 2 or tamano > 5:   # Validar el tamaño
    print("Noo es asi porque el tamaño debe estar entre 2 y 5.")
else:
    tabla = generar_tabla(tamano)    # Generar la tabla
    imprimir_tabla(tabla)            # Mostrar tabla

    renglon = int(input("Renglon (factor): "))   # Pedir factores
    columna = int(input("Columna (factor): "))

    if renglon < 1 or renglon > tamano or columna < 1 or columna > tamano:       # Validar los factores
        print("No, los factores deben estar dentro del tamaño de la tabla.")
    else:
        resultado = tabla[renglon - 1][columna - 1]     # Ver en la matriz
        print("El producto de", renglon, "x", columna, "es:", resultado)
```
