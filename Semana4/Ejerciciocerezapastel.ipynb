Ejercicio cereza en el pastel (diagrama de árbol con asteriscos pidiendo números del -10 al 10)

negativos izquierda, positivos derecha usando pipe

```python
numeros = []

for i in range(10):
    numero = int(input("Ingresa un número entre -10 y 10: "))

    while numero < -10 or numero > 10:
        print("Número no válido")
        numero = int(input("Ingresa un número entre -10 y 10: "))

    numeros.append(numero)
print("Gráfica:")

for numero in numeros:

    if numero < 0:
        print(" " * (10 - abs(numero)) + "*" * abs(numero) + "|")
    elif numero > 0:
        print(" " * 10 + "|" + "*" * numero)
    else:
        print(" " * 10 + "|")

# No se si esté bien profe, pero fue lo mejor que me quedó
```
