Uso de tuplas

1. Crea una tupla llamada "numeros" que contenga, al menos, cinco números.

2. Accede al tercer elemento de la tupla e imprímelo en pantalla.2

3. Por medio de una captura, agrega dos números adicionales a la tupla (emplea input).

4. Convierte la tupla en una lista y, después, ordena los elementos.

5. Crea una función que reciba la tupla de números como parámetro y retorne la suma de todos los elementos.

6. Aplica la función creada en el paso anterior a la tupla y, luego, muestra el resultado.

```python
numeros = (55, 67, 90, 9, 10)                   # Creación tupla
print("tercer numero:", numeros[2])             # Tercer número

numero1 = int(input("ingresa un numero adicional: "))     # Ingresar dos números nuevos
numero2 = int(input("ingresa el segundo numero adicional: "))

numeros = numeros + (numero1, numero2)

lista = list(numeros)
lista.sort()                                # Ordenar los números
print("numeros ordenados:", lista)

def suma_tupla(numeros):                      # Creación función para recibi tupla
    return sum(numeros)

resultado = suma_tupla(numeros)
print("la suuma de todos los elementos, es:", resultado)
```

Uso de diccionarios

1. Crea un diccionario llamado "contactos" con, al menos, tres entradas, donde las claves sean los nombres de las personas y los valores correspondan a sus números de teléfono.

2. Por medio de una captura, agrega un nuevo contacto al diccionario.

3. Itera sobre las claves del diccionario e imprime en pantalla los nombres de los contactos.

4. Crea una función que reciba el diccionario de contactos y un nombre como parámetro para que, posteriormente, retorne el número de teléfono correspondiente al contacto ingresado.

5. Aplica la función creada en el paso anterior para buscar el número de teléfono de uno de los contactos y muestra el resultado.

```python
contactos = {                        # Creación diccionario con tres entradas
    "Majo": "4427472819",
    "Aldo": "4422345678",
    "Messi": "442342193"
}

nombre = input("ingrese el nombre del nuevo contacto: ")      # Ingresar nuevo contaco y número
telefono = input("ingresa su numero de telefono: ")

contactos[nombre] = telefono
print("\ncontactos:")

for contacto in contactos:
    print(contacto)

def buscar_contacto(contactos, nombre):             # Creación función para recibir el diccionario de contactos
    return contactos.get(nombre, "el contacto no existe")

nombre_buscar = input("\ningrese el nombre del contacto que desee buscar: ")   # Buscar el contacto
resultado = buscar_contacto(contactos, nombre_buscar)
print(f"numero de telefono de {nombre_buscar}:", resultado)
```

Uso de excepciones

1. Crea un programa que solicite al usuario que ingrese dos números enteros.

2. Utiliza un bloque try-except para manejar la excepción que ocurrirá si el usuario ingresa un valor no numérico.

3. Si el usuario ingresa valores no numéricos, muestra un mensaje de error; en cambio, si los valores son numéricos, muestra la suma de los dos números ingresados.

4. Agrega una excepción personalizada para manejar la división entre cero; por tanto, en caso de que el segundo número ingresado sea cero, muestra un mensaje de error adecuado.

```python
num1 = input("ingrese el primer número: ")        # Definir primer número 
print (num1)
num2 = input("ingrese el segundo número: ")       # Definir segundo número
print (num2)

try:                                # Intento para valores no númercos
    if not num1.isdigit() or not num2.isdigit():
        print("ERRROR: debes ingresar solo números enteros.")
    elif num2 == "0":
        print("ERROR: el segundo numero no puede ser cero")   # División entre cero
        
    else:
        num1 = int(num1)
        num2 = int(num2)
        print("la suma de ambos numeross es:", num1 + num2)

except Exception:       
    print("ocurrio un error, favor de intentarlo de nuevo!!")
```

Uso de strings

1. Crea una variable llamada "mensaje" que contenga un texto de tu elección.

2. Imprime en pantalla la longitud del mensaje.

3. Utiliza un método de strings para convertir todo el texto del mensaje a mayúsculas.

4. Utiliza otro método de strings para reemplazar una palabra específica en el mensaje por otra de tu elección.

5. Crea una función que reciba un string como parámetro y retorne la cantidad de palabras que contiene.

6. Aplica la función creada en el paso anterior al mensaje y muestra el resultado.

```python
mensaje = "Messi es el mejor futbolista de todos los tiempos G.O.A.T"  # Creación variable mensaje
print("longitud del mensaje:", len(mensaje))

mensaje = mensaje.upper()                       # Mensaje en mayúsculas
print("mensaje en mayusculas:", mensaje)

mensaje1 = mensaje.replace("G.O.A.T", "el mejor de Argentina")          # Reemplazo de palabras
print("mensaje modificado:", mensaje1) 

def contar_palabras(texto):                  # Creación función para recibir string
    return len(texto.split())

cantidad = contar_palabras(mensaje)
print("cantidad de palabras:", cantidad)
```

Menú principal

1. Crea un menú principal que permita elegir qué tipo de dato se desea capturar.

2. De acuerdo con la elección hecha por el usuario, llama a la función indicada para proceder a su ejecución.

3. Agrega una opción para finalizar la ejecución.

```python
def dame_desayuno():                  # Creación función del desayuno
    print("pan francés, café, fruta & huevo")

def dame_comida():                    # Creación función de la comida
    print("Pechuga de pollo con guacamole, arroz & tortillas")

def dame_cena():                      # Creación función de lña cena
    print("Crema de verdura con 2 quesadillas")

opcion = ()                       # Menu
while opcion != "4":
    print("\nMENÚ")
    print("1. opcion del dia del desayuno")
    print("2. opcion del dia de la comida")
    print("3. opcion del dia de la cena")
    print("4. no tengo hambre!")
    opcion = input("ingrese la opcion que desee comprar: ")

    if opcion == "1":               # Llamar al cada función dependiendo la elección
        dame_desayuno()
    elif opcion == "2":
        dame_comida()
    elif opcion == "3":
        dame_cena()
    elif opcion == "4":
        print("el programa termina aquí")
    else:
        print("opcion incorrecta, vete a otro lugar si quieres")
```
