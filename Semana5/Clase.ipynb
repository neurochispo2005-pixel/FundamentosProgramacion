```python
def suma_tupla(t):
    total = 0
    for e in t:
        total += e
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)
print(f"Metiste a tu cuerpo {resultado} calorias, gordo")
```

# Diccionarios

```python
contactos = {
    "Ricardo": "44532235",  "Pedro": "44224959",
        "Luisa": "674838282"
}

contactos["Cris"]="299338843"
contactos["Ricardo"] = "384855847"

print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
print("Impresión de llaves")
for n in contactos.keys():
    print(n)

print("\nImpresion de valores")
for v in contactos.values():
    print(v)

print("\nImpresión completa")
for n, t in contactos.items():
    print(f"Nombre: {n}, Teléfono: {t}")
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")

nombre = input("Escribe el nombre del contacto que buscas")
telefono = buscar_contacto(contactos, nombre)
print(f"El telefono de {nombre} es: {telefono}")
```

```python
try:
    edad = int(input("Ingresa la edad:"))
    print(f"Tu edad es {edad}")
except ValueError:
    print("Error: Debes ingresar solo numeros enteros")
```

```python
try:
    dividendo = int(input("Dame un numero:"))
    divisor = int(input("Entre que numero lo dividimos"))
    resultado = dividendo/divisor
except ZeroDivisionError:
    print("Error de division entre cero")
else:
    print(f"El resultado es {resultado}")
finally:
    print("Fue un gusto hacer esa división")
```

```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("Ingrese la cantidad de votos:"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A,B o C):")
    if voto.upper() in candidatos:
        print(f"Voto para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es valido, vota nuevamente")

print("\nResultados:")
for k, v in candidatos.items():
    print(f"Opcion {k}, votos {v}")
```
