Desafío extra 1

Crea una tupla `espera = ("María", "José", "Carlos", "Lucía", "Pedro")`. Imprime el
tercer elemento, captura dos nombres nuevos con `input()`, anéxalos en una nueva
tupla, conviértela en lista, ordénala alfabéticamente y muestra la cantidad total
de personas.

```python
espera = ("María", "José", "Carlos", "Lucía", "Pedro")

print(espera[2])                            # Imprimir el tercer elemento

nombre1 = input("Ingresa un nombre nuevo: ")      # Ingresar dos nombres nuevos
nombre2 = input("Ingresa otro nombre nuevo: ")

espera = espera + (nombre1, nombre2)           # Anexarlos creando una nueva tupla

espera_lista = list(espera)                    # Convertir la tupla en lista
espera_lista.sort()                            # Ordenar alfabéticamente o en orden

print("Personas ordenadas:", espera_lista)                 # Mostrar la lista y la cantidad total
print("Cantidad total de personas:", len(espera_lista))
```

Desafío extra 2

Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

```python
candidatos = {                                   # Creación de diccionario
    "R": 0,
    "A": 0,
    "V": 0
}

cantidad = int(input("Ingrese la cantidad de votos:"))      # Pedir cantidad de votos
conteo = 1

while conteo <= cantidad:                     # Condición de votos aceptados
    voto = input(f"Ingresa el voto {conteo} (R, A o V):")
    if voto.upper() in candidatos:                  # Voto en mayúsculas
        print(f"Voto para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es valido, vota nuevamente")    # Voto fuera de los candidatos

print("\nResultados:")
for k, v in candidatos.items():           # Impresión con items()
    print(f"Opcion {k}, votos {v}")       # Impresión de cantidad de votos para cada color
```

Desafío extra 3

Escribe un programa que pida al usuario dos números enteros dentro de un bloque
`try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje
controlado y repite la solicitud con un ciclo `while`. Luego pide la división del
primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
while True:         # Repetir solicitud hasta que el usuario ponga dos números válidos
    try:
        numero1 = int(input("Ingresa el primer número entero: "))     # Pedimos los números
        numero2 = int(input("Ingresa el segundo número entero: "))
        break                       # Terminamos si son válidos numero1 y numero2
    except ValueError:              # Detectar el error
        print("No es válido. Debes ingresar números enteros y no dejar el campo vacío.")

try:                                    # Intentar hacer la división
    resultado = numero1 / numero2
    print(f"El resultado de la división es: {resultado}")
except ZeroDivisionError:               # Error en división entre cero
    print("Error: no se puede dividir entre cero.")
```

Desafío extra 4

Crea un programa que reciba una frase, imprima su longitud con `len()`, la
convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre
cuántas palabras contiene usando una función que retorne el conteo

```python
def contar_palabras(frase):            # Crear función contar_palabras
    return len(frase.split())          # Separar frase y nos indica cantidad

frase = input("Ingresa una frase: ")    # Ingresar frase
print("Longitud de la frase:", len(frase))

frase_mayusculas = frase.upper()        # La frase se vuelve en mayúsculas
print("En mayúsculas:", frase_mayusculas)

frase_reemplazada = frase.replace("Python", "programación")      # Remplazo de palabras
print("Con reemplazo:", frase_reemplazada)
print("Cantidad de palabras:", contar_palabras(frase))
```
