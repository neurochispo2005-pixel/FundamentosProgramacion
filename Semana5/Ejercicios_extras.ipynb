Ejercicio extra 1

Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)       # Creamos la tupla original con las calificaciones
print("Tercera calificación es:", calificaciones[2]) # Imprimir la tercera calificación

nueva1 = float(input("Nueva calificación 1: "))    # Pedimos dos nuevas calificaciones
nueva2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (nueva1, nueva2)     # Creamos una nueva tupla agregando las dos calificaciones
print("Nueva tupla:", nueva_tupla)               # Mostramos la nueva tupla

lista_calificaciones = list(nueva_tupla)            # Convertimos la nueva tupla en una lista

lista_calificaciones.sort(reverse=True)                # Ordenamos la lista de mayor a menor

print("Lista ordenada (mayor a menor):", lista_calificaciones)     # Mostramos la lista ordenada

def sumar_calificaciones(lista):                      # Creaciónfunción para calcular la suma de las calificaciones
    return sum(lista)

suma_total = sumar_calificaciones(lista_calificaciones)     # Llamamos a la función y mostramos el resultado
print("Suma total:", suma_total)
```

Ejercicio extra 2

Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.

```python
agenda = {                                  # Creación diccionario con tres contactos iniciales
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nombre_nuevo = input("Nombre del nuevo contacto: ")  # Nombre del nuevo contacto
telefono_nuevo = input("Teléfono del nuevo contacto: ")  # Teléfono del nuevo contacto

agenda[nombre_nuevo] = telefono_nuevo             # Agregamos el nuevo contacto al diccionario

nombres = agenda.keys()                      # Obtenemos todos los nombres del diccionario usando keys()

print("Contactos registrados:", ", ".join(nombres))

def buscar_telefono(nombre):             # Creación función buscar_telefono
    return agenda.get(nombre)          # get() busca el nombre en el diccionario y retorna su teléfono

nombre_buscar = input("Nombre a buscar: ")         # Nombre del contacto que queremos buscar
telefono = buscar_telefono(nombre_buscar)
print(f"El teléfono de {nombre_buscar} es: {telefono}")   # Mostramos el teléfono encontrado
```

Ejercicio extra 3

Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.

```python
try:                               # Intentamos pedir los dos números dentro del bloque try
    numero1 = int(input("Primer número entero: "))  # Pedimos lods dos numeros
    numero2 = int(input("Segundo número entero: "))

    resultado = numero1 / numero2       # Realizamos la división
    print(f"{numero1} dividido entre {numero2} es: {resultado}")

except ValueError:                              # Capturamos el error cuando se ingresan letras
    print("Error: debes ingresar números enteros válidos.")

except ZeroDivisionError:                  # Capturamos el error cuando el segundo número es 0
    print("Error: no es posible dividir entre cero.")
```

Ejercicio extra 4

Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.

```python
mensaje = "Python es un lenguaje poderoso"      # Creación variable con el mensaje original

print("Longitud del mensaje:", len(mensaje))    # len() para obtener la cantidad de caracteres del mensaje

mensaje_mayusculas = mensaje.upper()            # upper() para convertir todo el mensaje a mayúsculas
print("En mayúsculas:", mensaje_mayusculas)

texto_reemplazado = mensaje.replace("Python", "programación")  # replace() para cambiar "Python" por "programación"
print("Texto reemplazado:", texto_reemplazado)  

def contar_palabras(texto):      # Creación función contar_palanras
    return len(texto.split())

total_palabras = contar_palabras(mensaje)          # Llamamos a la función
print("Palabras totales:", total_palabras)
```
