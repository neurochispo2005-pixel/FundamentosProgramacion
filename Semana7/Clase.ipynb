```python
# 1. Modo 'w' borra el contenido previo y escribe "Dato original"
with open("datos.txt", "w") as archivo:
    archivo.write("Dato original\n")

# 2. Modo 'w' vuelve a sobreescribir todo
with open("datos.txt", "w") as archivo:
    archivo.write("Sobrescrito\n")

# 3. Modo 'w' sobreescribe una vez más
with open("datos.txt", "w") as archivo:
    archivo.write("Nuevo valor\n")

# 4. Modo 'a' anexa sin borrar lo existente
with open("datos.txt", "a") as archivo:
    archivo.write("Adicional\n")
```

Ejercicio 7

```python
numero = int(input("Ingresa un numero entero: "))

nombre_archivo = f"tabla_{numero}.txt"

with open(nombre_archivo, "w") as archivo:
    for i in range(1, 11):
        archivo.write(f"{numero} x {i} = {numero * i}\n")

print(f"Tabla guardada en {nombre_archivo}.")
```
